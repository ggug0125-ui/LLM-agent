```python
from langchain_openai import OpenAIEmbeddings 
from dotenv import load_dotenv
import os

load_dotenv()

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

embedding = OpenAIEmbeddings(
    model="text-embedding-3-large",
    api_key=OPENAI_API_KEY
)

v = embedding.embed_query("수원시 도시기본계획의 핵심 내용은 뭐야?")
print(len(v))
```

```python
from langchain_chroma import Chroma
import os

persist_directory = "../chroma_store_suwon"

# 저장된 수원 전용 크로마 DB가 없다면 새로 만들기
if not os.path.exists(persist_directory):
    print("Creating new Suwon Chroma store")
    vectorstore = Chroma.from_documents(
        documents=suwon_splits,
        embedding=embedding,
        persist_directory=persist_directory
    )

else:
    print("Loading existing Suwon Chroma store")
    vectorstore = Chroma(
        persist_directory=persist_directory,
        embedding_function=embedding
    )
```

```python
retriever = vectorstore.as_retriever(k=3)
docs = retriever.invoke("서울시의 환경 정책에 대해 궁금해")

for d in docs:
    print(d)
    print('------')
```

```python
# 사용자가 질문하면 리트리버를 이용하여 관련 내용을 찾고 이를 바탕으로 언어 모델을 사용해 답변을 생성하는 기능 추가
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder 
# ChatPromptTemplate 언어모델에 전달할 프롬프트를 정의하는 템플릿 클래스(역할, 메시지 설정)
# MessagesPlaceholder 사용자 메시지와 같은 동적 데이터를 프롬프트에 삽입하기 위해 사용하는 클래스(여러 메시지 입력 처리용)
# 신형 코드로 변경 from langchain.chains.combine_documents import create_stuff_documents_chain 
# 여러 텍스트를 결합해 답변 생성 체인을 만드는 함수 (기존 프롬프트를 기반으로 언어 모델을 통해 최종 답변 제공)
from langchain_openai import ChatOpenAI 
# chat이라는 객체 생성용 (져럼한 gpt-4o-mini 사용 예정 )

chat = ChatOpenAI(model="gpt-4o-mini") 

# ④ 답변을 생성할 때 사용할 프롬프트 템플릿 생성
question_answering_prompt = ChatPromptTemplate.from_messages(
    [
        ( # ⑤ 시스템에 역할 지정
            "system",
            "사용자의 질문에 대해 아래 context에 기반하여 답변하라.:\n\n{context}",
        ),
        MessagesPlaceholder(variable_name="messages"), # ⑥ 자리자 표시자로 설정 나눈 메시지를 동적으로 삽입함.
        # 이를 통해 사용자가 한 질문을 바탕으로 답변을 생성한다.
    ]
)

# 구형 방식 : document_chain = create_stuff_documents_chain(chat, question_answering_prompt) 
# 최신 방식 체인 (create_stuff_documents_chain 대신)
document_chain = question_answering_prompt | chat
# 검색된 문서 청크와 사용자 질문을 바탕으로 언어 모델을 호출하고 답변을 생성하는 역할을 한다.
```

```python
# 채팅 메시지를 저장하는 객체 사용
from langchain.memory import ChatMessageHistory

# 채팅 메시지 저장할 메모리 객체 생성
chat_history = ChatMessageHistory() 
# 사용자 질문을 메모리에 저장
chat_history.add_user_message("서울시의 온실가스 저감 정책에 대해 알려줘.") 

# 문서 검색하고 답변 생성
answer = document_chain.invoke(
    {
        "messages": chat_history.messages,
        "context": docs,
    }
)

# 생성된 답변 메모리에 저장
chat_history.add_ai_message(answer) 

print(answer)
```

```python
for m in chat_history.messages: # 저장된 대화 기록을 출력해 본다.
    print(m)
```

```python
# 질의 확장 구현하기
# 서울시 온실가스 저감 정책에 관한 대회에 이어서 뉴욕은? 질문시 해결법
# 이를 처리하려면 질문을 문맥에 맞게 확장하여 관련된 청크를 검색하고 답변을 생성해야 한다. -> 질의 확장
from langchain_core.output_parsers import StrOutputParser # 문자열 출력 파서를 불러옵니다.
```

```python
query_for_nyc = "뉴욕은?"

# query augmentation 
# 기존 대화 내용을 활용해 query_augmentation 수행
query_augmentation_prompt = ChatPromptTemplate.from_messages(
    [
        MessagesPlaceholder(variable_name="messages"), # 기존 대화 내용
        (
            "system",
            "기존의 대화 내용을 활용하여 사용자의 아래 질문의 의도를 파악하여 명료한 한 문장의 질문으로 변환하라. 대명사나 이, 저, 그와 같은 표현을 명확한 명사로 표현하라. :\n\n{query}",
        ),
    ]
)
```

```python
query_augmentation_chain = query_augmentation_prompt | chat | StrOutputParser()
```

```python
augmented_query = query_augmentation_chain.invoke({
    "messages": chat_history.messages,
    "query": query_for_nyc  
})

print(augmented_query)
```

```python
docs = retriever.invoke(augmented_query)

for d in docs:
    print(d)
    print('------')
```

```python
chat_history.add_user_message(query_for_nyc) # query_for_nyc에 "뉴욕은?" 추가

answer = document_chain.invoke(
    {
        "messages": chat_history.messages,
        "context": docs,
    }
)

# 생성된 답변 메모리에 저장
chat_history.add_ai_message(answer) 

print(answer)
```
